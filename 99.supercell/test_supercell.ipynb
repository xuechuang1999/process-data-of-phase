```python
from ase.io import read, write

# 1. 读取两个结构
atoms1 = read('POSCAR1', format='vasp')
atoms2 = read('POSCAR2', format='vasp')

# 2. 将第二个原胞内的所有原子在笛卡尔坐标系下沿 a 方向平移
# 平移距离为第一个原胞的 a 晶格常数
atoms2.positions[:, 0] += atoms1.cell[0, 0]

# 3. 将两个结构合并 (此时晶胞参数默认继承原 atoms1 的大小)
combined = atoms1 + atoms2

# 4. 更新合并后结构的晶胞参数 (将 a 方向的长度加倍)
combined.cell[0, 0] = atoms1.cell[0, 0] + atoms2.cell[0, 0]

# 5. 输出拼接后的新 POSCAR，并强制使用 Direct 坐标
write('POSCAR_combined', combined, format='vasp', vasp5=True, direct=True)
```
