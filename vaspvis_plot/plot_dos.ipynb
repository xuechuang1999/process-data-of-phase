```python
from vaspvis import standard
import os

phase_folder = ["1_c_fm-3m", "2_t_p42_nmc", "3_m_p21_c", "4_c-p-43m", "5_t-p4_nbm", "6_m_p21_m", "7_o_pbcn"]

doping_type_folder = ["h1z3o8", "h2z2o8", "h3z1o8"]

doping_site_folder = ["1-doping", "2-doping", "3-doping"]

dos_folder = 'pdos'

output_folder = "doping_pdos"

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

for i in range(len(phase_folder)):
    for j in range(len(doping_type_folder)):
        for k in range(len(doping_site_folder)):
            folder = os.path.join(phase_folder[i], doping_type_folder[j], doping_site_folder[k], dos_folder)
            if os.path.exists(folder):
                print(f"Processing folder: {folder}")
                standard.dos_elements_spin_polarized(
                    folder=folder,
                    elements=["Hf", "Zr", "O"],
                    energyaxis='x',
                    output=f"{output_folder}/{phase_folder[i]}--{doping_type_folder[j]}--{doping_site_folder[k]}--{dos_folder}",
                    save=True
                )
            else:
                print(f"Folder does not exist: {folder}, skipping.")

```
